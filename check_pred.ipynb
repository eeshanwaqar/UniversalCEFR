```python
from transformers import pipeline,AutoTokenizer, AutoModelForSequenceClassification


CEFR_LEVELS = ["A1", "A2", "B1", "B2"]
id2label = {i: lvl for i, lvl in enumerate(CEFR_LEVELS)}

# Load saved model and tokenizer
model_name = "eurobert_DA_model"
model = AutoModelForSequenceClassification.from_pretrained(model_name, trust_remote_code=True)
tokenizer = AutoTokenizer.from_pretrained(model_name,use_fast=True)

# Load pipeline
clf = pipeline("text-classification", model=model, tokenizer=tokenizer)
```

```python
def predict_cefr_label(text):
    raw = clf(text, truncation=True, max_length=256)[0]
    label_id = int(raw["label"].split("_")[-1]) 
    cefr = id2label[label_id]
    return {"cefr_level": cefr, "score": round(raw["score"], 4)}
```

```python
text="Edrych am rwyle arall i astudio? Dim problem"

result = predict_cefr_label(text)
print(f"Predicted CEFR level: {result['cefr_level']} ({result['score']})")
```
