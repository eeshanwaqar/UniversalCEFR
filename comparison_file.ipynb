```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# File paths – update if needed
welsh_bt_path = "files/welsh_back_translation.csv"
welsh_fr_bt_path = "files/welsh_back_translation_french.csv"
```

```python
# Load datasets
df_bt = pd.read_csv(welsh_bt_path)
df_bt_fr = pd.read_csv(welsh_fr_bt_path)
```

```python
# Add source label
df_bt["source"] = "Back Translation"
df_bt_fr["source"] = "French Back Translation"
```

```python
# Combine both for comparison
df_combined = pd.concat([df_bt, df_bt_fr], ignore_index=True)
```

```python
# Stats
group_stats = df_combined.groupby("source")[["BLEU", "chrF", "cosine_similarity"]].agg(["mean", "std", "min", "max", "count"])
display(group_stats)
```
