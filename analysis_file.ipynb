```python
import pandas as pd
```

```python
# Load your dataset
df = pd.read_csv("files/fineTuned_filtered_high_similarity.csv")
```

```python
# Compare CEFR levels
df['cefr_match'] = df['back_translated_english_predicted_cefr'] == df['cefr_level']
```

```python
# Count mismatches
mismatches = df[~df['cefr_match']]
num_mismatches = len(mismatches)
total = len(df)
```

```python
#Print summary
print(f"Total rows: {total}")
print(f"Number of mismatches: {num_mismatches}")
print(f"Mismatch rate: {num_mismatches / total:.2%}")
```
